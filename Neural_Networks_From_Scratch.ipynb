## 6. Softmax Activation

```python
# Implementing Softmax function from scratch
output_layer = [2.2,3.5,7.5,8.8]
def Softmax_Output(output):
    Softmax_Output_layer=[]
    total=0
    for i in output:
        total= total + np.exp(i)
    for i in output:
        value = np.exp(i)
        Softmax_Output_layer.append(value/total)
    return Softmax_Output_layer

Softmax_Output(output_layer)
```

```python
def cross_entropy_loss(y_pred, y_true):
    y_pred = np.clip(y_pred, 1e-7, 1 - 1e-7)
    return -np.sum(y_true * np.log(y_pred))
```

## 10. Binary Cross Entropy

```python
def binary_cross_entropy(y_true,y_pred):
    error = -(y_true*np.log(y_pred) + (1-y_true)*np.log(1-y_pred))
    return error
```

## 11. Derivatives

```python
def derived_mse(y_true,y_pred):
    n = len(y_true)
    error = (2/n)*(y_true-y_pred)
    return error
```

```python
def derived_bce(y_true,y_pred):
    error= (y_pred-y_true) / y_pred * (1-y_pred)
    return error
```

## 12. Backpropagation

```python
def backpropagation(y_pred, y_true, inputs):
    dz_dw = inputs
    dy_dz = y_pred * (1 - y_pred)
    dL_dy = 2 * (y_pred - y_true)
    dL_dw = dL_dy * dy_dz * dz_dw
    return dL_dw
```

## 13. Gradient Descent

```python
def gradient_descent(m,b,y_true,y_pred,inputs, learning_rate):
    n = len(y_true)
    dmse_dm = 2/n * np.sum((-inputs)*(y_true-y_pred))
    dmse_db = 2/n * np.sum((y_pred - y_true))
    m = m - learning_rate*dmse_dm
    b = b - learning_rate*dmse_db
    return m,b
```
